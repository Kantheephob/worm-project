```python
from worm_visualize_segmentation_test import get_dataset, classes_map, color_map
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
from PIL import Image
```

```python
labels = '/project/lt200264-saiwat/WormProject/data/worm-24022026/labels/worm-24022026-labels_7class_138images.csv'
roi_metadata = '/project/lt200264-saiwat/WormProject/data/worm-24022026/processed/cropped/dataset_meta.pkl'
masks_dir = '/project/lt200264-saiwat/WormProject/data/worm-24022026/processed/sam2_mask'
```

```python
dataset = get_dataset(labels, roi_metadata, masks_dir)
dataset
```

```python
df_img1 = dataset[dataset['stem'] == 'worm-000009'].copy()
# df_img1
```

```python
df_img2 = dataset[dataset['stem'] == 'worm-000001'].copy()
image = Image.open(df_img2.iloc[0]['raw_img_path']).convert('RGB')
image
```

```python
# Filter out Class 4 right from the start
cls_list = [cls for cls in df_img1['Class'].sort_values().unique() if cls in (0, 1, 2)]
cls_map = ['Young', 'Perfect', 'Old']

fig, axes = plt.subplots(1, len(cls_list), figsize=(3 * len(cls_list), 3))

for i, (cls, cls_map) in enumerate(zip(cls_list, cls_map)):
    # Select the first row matching this class
    row = df_img1[df_img1['Class'] == cls].iloc[0]
    
    # Load the raw image
    img = Image.open(row['raw_img_path'])
    
    # Extract [x, y, w, h] and convert to PIL's [left, top, right, bottom] format
    x, y, w, h = row['bbox']
    
    # Add optional padding (in pixels) to see some surrounding context
    pad = 20
    left = max(0, x - pad)
    top = max(0, y - pad)
    right = min(img.width, x + w + pad)
    bottom = min(img.height, y + h + pad)
    
    # Crop the image
    cropped_img = img.crop((left, top, right, bottom))
    
    # Handle single vs multiple subplot indexing
    ax = axes[i] if len(cls_list) > 1 else axes
    ax.imshow(cropped_img)
    ax.set_title(f"{cls_map}", fontsize=15)
    ax.axis('off')

plt.tight_layout()
plt.show()
```

```python
# 1. Count occurrences of each class and sort by class ID (0, 1, 2, ...)
class_counts = dataset['Class'].value_counts().sort_index()

# 2. Prepare labels and colors using your existing dictionaries
y_labels = [f"Class {cls}\n({classes_map.get(cls, 'Unknown')})" for cls in class_counts.index]
bar_colors = [mcolors.to_rgb(color_map.get(cls, 'gray')) for cls in class_counts.index]

# 3. Plot the horizontal bar chart
fig, ax = plt.subplots(figsize=(12, 7))
bars = ax.barh(y_labels, class_counts.values, color=bar_colors, edgecolor='black', alpha=0.85, height=0.6)

# 4. Add exact count labels to the right of each bar
for bar in bars:
    width = bar.get_width()
    ax.text(
        width + (0.015 * max(class_counts.values)),  # Slight offset to the right of the bar
        bar.get_y() + bar.get_height() / 2,          # Vertically centered on the bar
        f"{int(width):,}", 
        va='center', 
        ha='left', 
        fontsize=12, 
        fontweight='bold'
    )

# 5. Customize titles, grid, and layout
ax.set_title('Number of Instances per Class in Dataset', fontsize=18, fontweight='bold', pad=15)
ax.set_xlabel('Number of Instances', fontsize=14, labelpad=10)
ax.set_ylabel('Class', fontsize=14, labelpad=10)
ax.tick_params(axis='x', labelsize=11)
ax.tick_params(axis='y', labelsize=11)
ax.grid(axis='x', linestyle='--', alpha=0.5)  # Switched grid to X-axis

# Expand x-axis limit slightly so the number labels on the right don't get cut off
ax.set_xlim(0, max(class_counts.values) * 1.15)

# Optional: Invert Y-axis so Class 0 starts at the top
ax.invert_yaxis()

plt.tight_layout()
plt.show()
```
